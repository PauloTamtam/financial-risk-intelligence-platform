```python
# Import core libraries
# pandas: data manipulation
# os: file path handling
import pandas as pd
import os

# Path to the raw dataset directory
RAW_PATH = "../data/raw/"

# List of all CSV files included in the Home Credit dataset
files = [
    "application_train.csv",
    "application_test.csv",
    "bureau.csv",
    "bureau_balance.csv",
    "credit_card_balance.csv",
    "installments_payments.csv",
    "POS_CASH_balance.csv",
    "previous_application.csv"
]

print("📁 Raw data directory:", RAW_PATH)
```

```python
def load_csv(file_name):
    """
    Loads a CSV file from the raw data directory.

    Parameters:
        file_name (str): Name of the CSV file.

    Returns:
        pd.DataFrame: Loaded dataset.
    """
    path = os.path.join(RAW_PATH, file_name)
    df = pd.read_csv(path)

    # Print basic metadata for quick inspection
    print(f"✔ Loaded {file_name} — shape: {df.shape}")

    return df
```

```python
# Dictionary to store all loaded DataFrames
data = {}

# Loop through each file and load it
for f in files:
    df = load_csv(f)
    data[f.replace(".csv", "")] = df  # Store using clean key names

print("\n📦 Data collection completed. All files successfully loaded.")
```

```python
print("🔍 Checking dataset integrity...\n")

# Count missing values for each dataset
for name, df in data.items():
    missing = df.isnull().sum().sum()
    print(f"{name}: {missing} missing values")
```

```python
# Display the first rows of the main training dataset
data["application_train"].head()
```

```python
# Directory for processed/validated datasets
PROCESSED_PATH = "../data/processed/"

# Create directory if it doesn't exist
if not os.path.exists(PROCESSED_PATH):
    os.makedirs(PROCESSED_PATH)

# Save validated copies of each dataset
for name, df in data.items():
    df.to_csv(os.path.join(PROCESSED_PATH, f"{name}_validated.csv"), index=False)

print("💾 Validated datasets saved to data/processed/")
```
